```python
import numpy as np
import pandas as pd
```

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
statlog_heart = fetch_ucirepo(id=145) 
  
# data (as pandas dataframes) 
X = statlog_heart.data.features 
y = statlog_heart.data.targets 
  
# metadata 
print(statlog_heart.metadata) 
  
# variable information 
print(statlog_heart.variables) 
```

```python
dataset=pd.read_csv("datasets/statlog+heart\heart.dat",sep=" ")
dataset
print(dataset.shape)
```

```python
dataset.isna().sum()
```

```python
x=dataset.iloc[:,:-1]
x.shape
y=dataset.iloc[:,-1:]
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape,y_train.shape,x_test.shape,y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
log=LogisticRegression(solver='newton-cg')
log.fit(x_train,y_train)
y_pred=log.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
cm=confusion_matrix(y_test,y_pred)
print(classification_report(y_test,y_pred))
print(cm)
```
